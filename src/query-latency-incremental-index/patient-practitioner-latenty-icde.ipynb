```python
import numpy as np
import matplotlib.pyplot as plt

from matplotlib.lines import Line2D
from matplotlib.patches import FancyBboxPatch
from matplotlib.ticker import LogLocator, LogFormatterMathtext

# ============================================================
# ESPRESSO FIGURE 3
# Colourful ICDE paper figure
#
# Four strategies:
#
# Baseline     : Deep Teal
# Case 1       : Golden Yellow
# Case 2       : Coral
# Case 3 (HDT) : Purple
#
# Metric:
# Mean end-to-end query latency
#
# Original data: milliseconds
# Plotted data : seconds
#
# Workload:
# Patient queries      Q01-Q04
# Practitioner queries Q05-Q08
#
# Federation sizes:
# 100, 250, 500, 750, 1059 Pods
# ============================================================


# ============================================================
# FEDERATION SIZES
# ============================================================

pods = [100, 250, 500, 750, 1059]

x = np.arange(len(pods))

queries = [
    "Q01", "Q02", "Q03", "Q04",
    "Q05", "Q06", "Q07", "Q08"
]

strategies = [
    "Baseline",
    "Case 1",
    "Case 2",
    "Case 3"
]


# ============================================================
# ACTUAL EXPERIMENTAL DATA
#
# Source:
# Latest experimental results supplied by the user.
#
# Column used:
# Total (ms)
#
# Values are mean end-to-end latency in milliseconds.
#
# Each list follows:
# [100, 250, 500, 750, 1059]
# ============================================================

data = {

    # ========================================================
    # PATIENT QUERY Q01
    # ========================================================

    "Q01": {

        "Baseline": [
            13433.645,
            32561.1736,
            64711.60515,
            88271.40875,
            127569.5717
        ],

        "Case 1": [
            2853.785,
            6670.76855,
            9357.52255,
            16133.4474,
            19571.3265
        ],

        "Case 2": [
            3730.9757,
            9320.86445,
            18024.8064,
            23847.56805,
            28775.49985
        ],

        "Case 3": [
            5004.65165,
            13470.9721,
            18756.22295,
            32765.39735,
            39846.15275
        ]
    },


    # ========================================================
    # PATIENT QUERY Q02
    # ========================================================

    "Q02": {

        "Baseline": [
            12983.9316,
            32793.09075,
            56932.46655,
            82441.3332,
            145474.4404
        ],

        "Case 1": [
            2348.50935,
            6600.8407,
            9031.87725,
            16576.51885,
            20142.2559
        ],

        "Case 2": [
            3201.9838,
            9457.28755,
            17443.6416,
            24986.5924,
            30466.28205
        ],

        "Case 3": [
            4697.3241,
            13905.26185,
            19541.4245,
            35863.6982,
            44501.12635
        ]
    },


    # ========================================================
    # PATIENT QUERY Q03
    # ========================================================

    "Q03": {

        "Baseline": [
            12030.5632,
            35440.90875,
            60830.2518,
            102285.6904,
            144771.181
        ],

        "Case 1": [
            2334.94075,
            9181.3036,
            15164.84745,
            32235.0762,
            47163.74615
        ],

        "Case 2": [
            3173.81355,
            12185.04125,
            23156.67885,
            41376.74715,
            57231.68565
        ],

        "Case 3": [
            4775.00725,
            17344.0412,
            25913.08595,
            55163.6971,
            77537.6962
        ]
    },


    # ========================================================
    # PATIENT QUERY Q04
    # ========================================================

    "Q04": {

        "Baseline": [
            12050.7297,
            36100.2325,
            59902.17125,
            92238.8088,
            157165.0176
        ],

        "Case 1": [
            2419.82445,
            6878.0211,
            9457.32405,
            17109.0547,
            21023.97395
        ],

        "Case 2": [
            3783.22515,
            12059.0109,
            18396.67465,
            30300.18515,
            36665.8592
        ],

        "Case 3": [
            9387.3525,
            30289.5315,
            43523.4655,
            80605.8172,
            98847.4033
        ]
    },


    # ========================================================
    # PRACTITIONER QUERY Q05
    # ========================================================

    "Q05": {

        "Baseline": [
            10902.07805,
            30921.9094,
            46873.65945,
            63923.7624,
            157919.7892
        ],

        "Case 1": [
            1077.33705,
            1395.4933,
            1447.4632,
            1525.9856,
            1990.84605
        ],

        "Case 2": [
            1133.3624,
            1503.17375,
            1615.7998,
            2035.92805,
            2687.42145
        ],

        "Case 3": [
            1908.28375,
            2736.32315,
            2987.39425,
            3396.85325,
            4863.06915
        ]
    },


    # ========================================================
    # PRACTITIONER QUERY Q06
    # ========================================================

    "Q06": {

        "Baseline": [
            11582.322,
            26310.39295,
            56003.62845,
            68097.4388,
            109438.9559
        ],

        "Case 1": [
            956.8792,
            1289.9871,
            1346.8233,
            1464.7612,
            1634.1678
        ],

        "Case 2": [
            1040.2893,
            1597.3957,
            1834.00035,
            2292.06535,
            3533.81575
        ],

        "Case 3": [
            1862.7303,
            2667.4857,
            3166.2246,
            3815.42765,
            5908.57375
        ]
    },


    # ========================================================
    # PRACTITIONER QUERY Q07
    # ========================================================

    "Q07": {

        "Baseline": [
            10712.5602,
            52472.0263,
            52105.1212,
            72066.35355,
            118018.5341
        ],

        "Case 1": [
            964.23375,
            1290.8913,
            1343.809,
            1405.77055,
            1527.94705
        ],

        "Case 2": [
            1092.3494,
            1826.1219,
            2246.31935,
            2483.71365,
            3587.3366
        ],

        "Case 3": [
            1903.0382,
            2934.84785,
            3589.0004,
            4023.5944,
            6144.40645
        ]
    },


    # ========================================================
    # PRACTITIONER QUERY Q08
    # ========================================================

    "Q08": {

        "Baseline": [
            11100.16695,
            39794.74735,
            54543.1885,
            80671.72775,
            379308.1889
        ],

        "Case 1": [
            1919.43535,
            2972.61625,
            4675.84765,
            6497.33425,
            36373.62345
        ],

        "Case 2": [
            2124.26615,
            4051.4915,
            8755.82935,
            9667.03605,
            212575.4385
        ],

        "Case 3": [
            17061.7539,
            39896.0751,
            78816.5544,
            115635.8616,
            398340.1348
        ]
    }

}


# ============================================================
# CONVERT MILLISECONDS TO SECONDS
#
# Formula:
#
# seconds = milliseconds / 1000
#
# No rounding is applied before plotting.
# ============================================================

data_seconds = {}

for query in queries:

    data_seconds[query] = {}

    for strategy in strategies:

        data_seconds[query][strategy] = (
            np.array(
                data[query][strategy],
                dtype=float
            ) / 1000.0
        )


# ============================================================
# VALIDATE EXPERIMENTAL DATA
# ============================================================

for query in queries:

    for strategy in strategies:

        values = data_seconds[query][strategy]

        assert len(values) == len(pods), (
            f"Incorrect number of values: "
            f"{query}, {strategy}"
        )

        assert np.all(np.isfinite(values)), (
            f"Invalid values: {query}, {strategy}"
        )

        assert np.all(values > 0), (
            f"Logarithmic plots require positive values: "
            f"{query}, {strategy}"
        )


# ============================================================
# COLOUR PALETTE
# ============================================================

TEAL = "#007F86"
GOLD = "#F5B700"
CORAL = "#FF6757"
PURPLE = "#8055C8"

DARK_TEAL = "#063E4C"

GRID = "#A9D9E0"

HEADER_BG = "#D7F0F1"

BORDER = "#00868B"


# ============================================================
# STRATEGY STYLES
# ============================================================

styles = {

    "Baseline": {
        "color": TEAL,
        "marker": "o",
        "linestyle": "-",
        "linewidth": 2.8,
        "markersize": 7.5
    },

    "Case 1": {
        "color": GOLD,
        "marker": "s",
        "linestyle": "--",
        "linewidth": 2.8,
        "markersize": 7.5
    },

    "Case 2": {
        "color": CORAL,
        "marker": "^",
        "linestyle": "--",
        "linewidth": 2.8,
        "markersize": 8
    },

    "Case 3": {
        "color": PURPLE,
        "marker": "D",
        "linestyle": "-.",
        "linewidth": 2.8,
        "markersize": 7
    }

}


# ============================================================
# GLOBAL FONT AND EXPORT SETTINGS
# ============================================================

plt.rcParams.update({

    "font.family": "DejaVu Sans",

    "font.size": 12,

    "axes.titlesize": 15,

    "axes.labelsize": 15,

    "xtick.labelsize": 11,

    "ytick.labelsize": 12,

    "axes.linewidth": 0.9,

    "figure.facecolor": "white",

    "savefig.facecolor": "white",

    "svg.fonttype": "none",

    "pdf.fonttype": 42

})


# ============================================================
# CREATE FIGURE
#
# 2 rows x 4 columns
#
# Row 1: Patient Q01-Q04
# Row 2: Practitioner Q05-Q08
# ============================================================

fig, axes = plt.subplots(

    2,
    4,

    figsize=(14, 8.5),

    sharex=True,
    sharey=True

)

fig.patch.set_facecolor("white")


# ============================================================
# DRAW ALL EIGHT QUERY PANELS
# ============================================================

for i, query in enumerate(queries):

    ax = axes.flat[i]

    ax.set_facecolor("white")

    # --------------------------------------------------------
    # DRAW FOUR STRATEGIES
    # --------------------------------------------------------

    for strategy in strategies:

        latency_seconds = data_seconds[query][strategy]

        ax.plot(

            x,

            latency_seconds,

            **styles[strategy],

            markeredgecolor="white",

            markeredgewidth=0.7,

            solid_capstyle="round",

            zorder=5

        )

    # --------------------------------------------------------
    # LOGARITHMIC LATENCY AXIS
    # --------------------------------------------------------

    ax.set_yscale("log")

    # All observed values fit within this range.
    ax.set_ylim(0.7, 650)

    # Major log ticks
    ax.yaxis.set_major_locator(
        LogLocator(
            base=10.0,
            numticks=5
        )
    )

    ax.yaxis.set_major_formatter(
        LogFormatterMathtext(base=10.0)
    )

    # --------------------------------------------------------
    # X-AXIS
    # --------------------------------------------------------

    ax.set_xticks(x)

    ax.set_xticklabels(
        ["100", "250", "500", "750", "1,059"],
        rotation=0
    )

    # Show labels on both rows
    ax.tick_params(

        axis="x",

        labelbottom=True,

        labelsize=10,

        pad=6

    )

    ax.tick_params(

        axis="y",

        labelsize=11,

        pad=5

    )

    # --------------------------------------------------------
    # GRIDLINES
    # --------------------------------------------------------

    ax.grid(

        True,

        which="major",

        axis="both",

        color=GRID,

        linestyle="--",

        linewidth=1.0,

        alpha=0.95,

        zorder=0

    )

    ax.grid(
        False,
        which="minor"
    )

    ax.set_axisbelow(True)

    # --------------------------------------------------------
    # PANEL BORDER
    # --------------------------------------------------------

    for spine in ax.spines.values():

        spine.set_color(DARK_TEAL)

        spine.set_linewidth(0.9)

    # --------------------------------------------------------
    # QUERY TITLE
    # --------------------------------------------------------

    ax.text(

        0.5,

        1.12,

        query,

        transform=ax.transAxes,

        ha="center",

        va="center",

        fontsize=16,

        fontweight="bold",

        color=DARK_TEAL,

        bbox=dict(

            boxstyle="round,pad=0.35",

            facecolor=HEADER_BG,

            edgecolor="none"

        ),

        clip_on=False

    )


# ============================================================
# FIGURE SPACING
# ============================================================

fig.subplots_adjust(

    left=0.095,

    right=0.975,

    top=0.82,

    bottom=0.18,

    hspace=0.90,

    wspace=0.14

)


# ============================================================
# ROW HEADINGS
# ============================================================

fig.text(

    0.5,

    0.935,

    "(a) Patient queries (Q01–Q04)",

    ha="center",

    va="center",

    fontsize=20,

    fontweight="bold",

    color=DARK_TEAL,

    bbox=dict(

        boxstyle="round,pad=0.45",

        facecolor=HEADER_BG,

        edgecolor="none"

    )

)


fig.text(

    0.5,

    0.505,

    "(b) Practitioner queries (Q05–Q08)",

    ha="center",

    va="center",

    fontsize=20,

    fontweight="bold",

    color=DARK_TEAL,

    bbox=dict(

        boxstyle="round,pad=0.45",

        facecolor=HEADER_BG,

        edgecolor="none"

    )

)


# ============================================================
# SHARED AXIS LABELS
# ============================================================

fig.text(

    0.025,

    0.52,

    "End-to-end latency (s, log scale)",

    rotation=90,

    ha="center",

    va="center",

    fontsize=16,

    color="#111827"

)


fig.text(

    0.5,

    0.115,

    "Federation size (Pods)",

    ha="center",

    va="center",

    fontsize=16,

    color="#111827"

)


# ============================================================
# LEGEND
#
# All four strategies
# ============================================================

handles = []

for strategy in strategies:

    s = styles[strategy]

    handles.append(

        Line2D(

            [0],
            [0],

            color=s["color"],

            marker=s["marker"],

            linestyle=s["linestyle"],

            linewidth=s["linewidth"],

            markersize=s["markersize"],

            label=strategy

        )

    )


legend = fig.legend(

    handles=handles,

    loc="lower center",

    bbox_to_anchor=(0.5, 0.025),

    ncol=4,

    fontsize=14,

    frameon=True,

    fancybox=True,

    columnspacing=2.5,

    handlelength=3.0,

    borderpad=0.7

)


legend.get_frame().set_facecolor("white")

legend.get_frame().set_edgecolor("#8BAEB6")

legend.get_frame().set_linewidth(1.1)


# ============================================================
# DECORATIVE OUTER BORDER
# ============================================================

outer_border = FancyBboxPatch(

    (0.007, 0.007),

    0.986,

    0.986,

    transform=fig.transFigure,

    boxstyle="round,pad=0.003,rounding_size=0.015",

    linewidth=1.8,

    edgecolor=BORDER,

    facecolor="none",

    clip_on=False,

    zorder=20

)

fig.add_artist(outer_border)


# ============================================================
# SAVE HIGH-QUALITY FILES
# ============================================================

fig.savefig(

    "espresso_figure3_four_strategies.pdf",

    format="pdf",

    bbox_inches="tight",

    pad_inches=0.08

)


fig.savefig(

    "espresso_figure3_four_strategies.svg",

    format="svg",

    bbox_inches="tight",

    pad_inches=0.08

)


fig.savefig(

    "espresso_figure3_four_strategies.png",

    format="png",

    dpi=400,

    bbox_inches="tight",

    pad_inches=0.08

)


# ============================================================
# DISPLAY FIGURE
# ============================================================

plt.show()


# ============================================================
# DISPLAY LARGE PNG IN GOOGLE COLAB
# ============================================================

from IPython.display import display, Image

display(

    Image(

        filename="espresso_figure3_four_strategies.png",

        width=1400

    )

)


# ============================================================
# PRINT LATENCY VALUES IN SECONDS
#
# Useful for checking the plotted values.
# ============================================================

print("\n" + "=" * 80)

print("ESPRESSO FIGURE 3 — END-TO-END LATENCY (SECONDS)")

print("=" * 80)


for query in queries:

    print(f"\n{query}")

    print("-" * 80)

    for strategy in strategies:

        values = data_seconds[query][strategy]

        formatted = [
            f"{value:.3f}"
            for value in values
        ]

        print(
            f"{strategy:<15}: "
            + " | ".join(formatted)
        )


print("\n" + "=" * 80)

print("FIGURE FILES SAVED SUCCESSFULLY")

print("=" * 80)

print("PDF : espresso_figure3_four_strategies.pdf")

print("SVG : espresso_figure3_four_strategies.svg")

print("PNG : espresso_figure3_four_strategies.png")
```
